```python
# Do not change this cell in your submission
import numpy as np
X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X + np.random.randn(100, 1)
```

# Question 1 (1 mark)
**Q1**: You need to output the shapes of X and y in separate lines using numpy.
- Write your code in the cell below
- Do not delete the comment line # Your answer here

```python
# Q1 - Your answer here

X.shape
y.shape
```

We can use matplotlib to visualize the dataset:

```python
# Do not change this cell in your submission
import matplotlib.pyplot as plt
plt.plot(X, y, "b.")
plt.show()
```

We can scale the data using pandas.

```python
# Do not change this cell in your submission
import pandas as pd

data = {'n_comp_affected': X.flatten(), 'cost': y.flatten()}
df = pd.DataFrame(data)
df['n_comp_affected'] = df['n_comp_affected'] * 1000
df['n_comp_affected'] = df['n_comp_affected'].astype('int')
df['cost'] = df['cost'] * 1000
df['cost'] = df['cost'].astype('int')
```

# Question 2 (1 mark)
**Q2**: You need to randomly sample 5 rows from the DataFrame, and output the results as a DataFrame.

```python
# Q2 - Your answer here

df.sample(n=5)
```

# Question 3 (1 mark)
**Q3**: You need to output the minimum valule and the maximum value of the feature `n_comp_affected`.
- output them as an ndarray object named `X_tr`. 
- the shape of `X_tr` should be $2 \times 1$.

```python
# Q3 - Your answer here:

min = df['n_comp_affected'].min()
max = df['n_comp_affected'].max()
X_tr = np.array([[min],[max]])
X_tr.reshape((2, 1))
```

Now we can train the linear regression model using the dataset.

```python
# Do not change this cell in your submission
from sklearn.linear_model import LinearRegression

# For simplicity, we are using all the data for training in this example, and evaluation is not considered.
lin_reg = LinearRegression()
lin_reg.fit(df['n_comp_affected'].values.reshape(-1, 1), df['cost'].values)
```

# Question 4 (1 mark)
**Q4**: You need to use your model to compute the target value (the 'cost' column) of `X_tr` and output the prediction.
- output the prediction as a 1-d array named `y_pred_tr`

```python
# Q4 - Your answer here

y_pred_tr = lin_reg.predict(X_tr).flatten()
```

# Question 5 (1 mark)
**Q5**: You need to plot the dataset and your linear regression model in the same figure using matplotlib.

- show your name and student ID above the figure
- use BLUE DOTS to represent the training data
- use RED LINE to represent the model

```python
# Q5 - Your answer here

plt.plot(df['n_comp_affected'], df['cost'], "b")
plt.plot(X_tr, y_pred_tr, "r")
plt.show()
```
